## 1. 데이터 로딩

```python
# 따릉이 데이터 로딩
bike = load_bike_data(data_dir="../data/raw")
bike.head()
```

```python
# 대여소 마스터 로딩
station = load_station_master(file_path="../data/external/서울시 따릉이대여소 마스터 정보.csv")
station.head()
```

```python
# 기상 데이터 로딩
weather = load_weather_data(file_path="../data/external/서울시 지상관측자료 정보(일별).csv")
weather.head()
```

## 2. 데이터 구조 확인

```python
# 기상 데이터 정보
print("=== 기상 데이터 구조 ===")
weather.info()
```

## 3. 결측값 확인

```python
check_missing_values(bike, "따릉이 데이터")
```

```python
check_missing_values(station, "대여소 마스터")
```

```python
check_missing_values(weather, "기상 데이터")
```

## 4. 기본 통계

```python
# 따릉이 이용 통계
bike[['전체_건수', '전체_이용_분', '전체_이용_거리']].describe()
```

```python
# 기상 통계
weather[['평균기온(℃)', '일강수량(mm)', '평균풍속(m/s)', '평균상대습도(%)']].describe()
```

## 5. 시간대별 분포

```python
# 시간대별 이용 건수
hourly_dist = bike.groupby('기준_시간')['전체_건수'].sum().sort_index()

plt.figure(figsize=(12, 5))
plt.bar(hourly_dist.index, hourly_dist.values, color='steelblue')
plt.title('시간대별 총 이용 건수', fontsize=14, fontweight='bold')
plt.xlabel('시간대 (시)', fontsize=12)
plt.ylabel('총 이용 건수', fontsize=12)
plt.grid(axis='y', alpha=0.3)
plt.show()
```

## 6. 대여소별 이용 분포

```python
# 상위 20개 대여소
top_stations = bike.groupby('시작_대여소_ID')['전체_건수'].sum().sort_values(ascending=False).head(20)

plt.figure(figsize=(12, 6))
plt.barh(range(len(top_stations)), top_stations.values, color='coral')
plt.yticks(range(len(top_stations)), top_stations.index)
plt.xlabel('총 이용 건수', fontsize=12)
plt.ylabel('대여소 ID', fontsize=12)
plt.title('이용량 상위 20개 대여소', fontsize=14, fontweight='bold')
plt.gca().invert_yaxis()
plt.grid(axis='x', alpha=0.3)
plt.show()
```
